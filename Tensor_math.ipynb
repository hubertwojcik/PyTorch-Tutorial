```python
import torch
import numpy as np
```

# Addition

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])

tensor_a + tensor_b
```

```python
# Addition longhand
torch.add(tensor_a, tensor_b)
```

# Subtraction

```python
tensor_b - tensor_a
```

```python
# Sub function
torch.subtract(tensor_b, tensor_a)
```

# Multiplication

```python
tensor_a * tensor_b
```

```python
# Multiplicaiton longhand
torch.mul(tensor_a, tensor_b)
```

# Division

```python
tensor_b / tensor_a
```

```python
# Division longhand
torch.divide(tensor_a, tensor_b)
```

# Modulo

```python
tensor_b % tensor_a
```

# Power

```python
tensor_a ** tensor_b
```

# Reassignment

```python
tensor_a.add(tensor_b)
```
